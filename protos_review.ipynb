## Interactive Exploration

Now that we've seen the complete framework, let's do some interactive exploration:

````python
# Explore the data directory structure
import os

data_dir = review.data_dir
print(f"📁 Data Directory: {data_dir}")
print("\n📂 Directory Structure:")

for root, dirs, files in os.walk(data_dir):
    level = root.replace(str(data_dir), '').count(os.sep)
    indent = ' ' * 2 * level
    rel_path = os.path.relpath(root, data_dir)
    if rel_path == '.':
        rel_path = 'data/'
    print(f"{indent}{rel_path}/")
    
    # Show first few files in each directory
    sub_indent = ' ' * 2 * (level + 1)
    for file in files[:3]:  # Limit to first 3 files
        print(f"{sub_indent}{file}")
    if len(files) > 3:
        print(f"{sub_indent}... and {len(files) - 3} more files")
````

````python
# Explore the entity registry
global_registry = review.components['GlobalRegistry']()
all_entities = global_registry.entity_registry.list_entities()

print(f"🔍 Entity Registry Exploration")
print(f"   Total entities: {len(all_entities)}")

# Analyze format distribution
format_stats = {}
for entity_id in all_entities[:10]:  # Look at first 10 entities
    entity_info = global_registry.entity_registry.get_entity(entity_id)
    if entity_info:
        original_id = entity_info.get('original_id', 'Unknown')
        formats = list(entity_info.get('formats', {}).keys())
        
        print(f"\n   Entity: {entity_id}")
        print(f"      Original ID: {original_id}")
        print(f"      Formats: {formats}")
        
        for fmt in formats:
            format_stats[fmt] = format_stats.get(fmt, 0) + 1

print(f"\n📊 Format Distribution (first 10 entities):")
for fmt, count in sorted(format_stats.items()):
    print(f"   {fmt}: {count} entities")
````

````python
# Try loading data from different processors
print("🧪 Cross-Processor Data Access Demo")

# If we have a sequence processor, show some sequences
if 'sequence' in processors:
    seq_proc = processors['sequence']
    entities = seq_proc.list_entities()
    
    print(f"\n🧬 Sequence Processor: {len(entities)} entities")
    for entity in entities[:3]:
        sequence = seq_proc.get_sequence(entity)
        if sequence:
            print(f"   {entity}: {sequence[:30]}... ({len(sequence)} residues)")

# If we have a structure processor, show some structures
if 'structure' in processors:
    struct_proc = processors['structure']
    entities = struct_proc.list_entities()
    
    print(f"\n🏗️ Structure Processor: {len(entities)} entities")
    for entity in entities[:3]:
        try:
            data = struct_proc.load_structure(entity)
            if data is not None:
                chains = data['auth_chain_id'].unique()
                print(f"   {entity}: {len(data)} atoms, chains {sorted(chains)}")
        except Exception as e:
            print(f"   {entity}: Could not load ({e})")

# If we have a property processor, show some properties
if 'property' in processors:
    prop_proc = processors['property']
    datasets = prop_proc.list_datasets()
    
    print(f"\n📊 Property Processor: {len(datasets)} datasets")
    for dataset in datasets[:2]:
        dataset_id = dataset['id']
        stats = prop_proc.get_dataset_statistics(dataset_id)
        print(f"   {dataset_id}: {stats['entity_count']} entities, {stats['property_count']} properties")
````
